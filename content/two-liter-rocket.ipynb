## Two Liter Rocket

Use conservation of momentum to calculate exit velocity of fuel.

```python
from IPython.display import display, Math, Latex
from math import log
display(Latex(r'$$ m_i v_i = m_r v_r + m_w v_w $$'))
m_r = 0.5
m_w = 0.375
m_i = m_r + m_w
g = 10 # m/s2
v_r = 8 * g * 0.1

display(Latex(rf'$$ 0.875 \times 0 = 0.5 \times {v_r}  + 0.375 v_w $$'))
display(Latex(rf'$$ 0 = {0.5*v_r}  + 0.375 v_w $$'))
display(Latex(rf'$$ {0.5*v_r} = - 0.375 v_w $$'))
display(Latex(rf'$$ {0.5*v_r} / (-0.375) = v_w $$'))
v_e = -0.5*v_r/m_w
display(Latex(rf'$$ v_w = {"{:.2f}".format(v_e)} m/s $$'))

del_v = v_r 
# https://en.wikipedia.org/wiki/Tsiolkovsky_rocket_equation
del_v_actual = v_e * log(m_i / m_r)
print(del_v, del_v_actual)
```
